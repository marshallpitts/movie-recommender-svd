```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Importing libraries

```python
movies = ["Action 1", "Action 2", "Action 3",
"Romance 1", "Romance 2", "Romance 3"]
users = ["Ana", "Ben", "Cal", "Dee", "Eve", "Fay"]
```

Created two lists: movies and users to hold string values of different users and movies.

```python
toy = np.array([
[5, 4, 5, 1, np.nan, 2], # Ana
[4, np.nan, 5, 2, 1, 1], # Ben
[5, 5, np.nan, 1, 2, np.nan], # Cal
[1, 2, 1, 5, 4, np.nan], # Dee
[np.nan, 1, 2, 4, 5, 5], # Eve
[2, 1, np.nan, 5, np.nan, 4], # Fay
])
```

Ratings of the movies from the users, np.nan represents "not a number" meaning the user has no rating.

```python
toy_table = pd.DataFrame(toy, index=users, columns=movies)
toy_table
```

Created a 6x6 matrix where each row corresponds to the matching person in users and each column matches with a movie in movies.

```python
missing = np.isnan(toy).sum()
print("Missing entries:", missing, "out of", toy.size)
```

Create a true/false matrix where true holds if an entry is missing. Then those values are counted and are given out of the total number of entries.
